```python
# Enunciado: Imprime los números del 1 al 15 haciendo servir un bucle while, pero salta/termina el bucle inmediatamente si el número llega a 10.

num = 1
while num <=15:
    if num == 10:
        break
    print("num:",num)
    num +=1
```

```python
# Enunciado: Calcula la suma de los números impares entre 1 y 20 (inclusos) haciendo servir un bucle while.
num = 1
impar = 0

while num <= 20:
    if num % 2 != 0:
        impar += num
    num +=1
print(f"Suma de impares: {impar}")
```

```python
# Enunciado: Demana a l'usuari que introdueixi un número enter positiu. Calcula i imprimeix la suma de totes les seves xifres fent servir un bucle while.
numero = int(input("Introdueix un número enter positiu: "))
suma_digits = 0
while numero > 0:
    suma_digits += numero % 10
    numero //= 10
print("La suma de les xifres és:", suma_digits)
```
